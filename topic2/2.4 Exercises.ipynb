Write a function that computes the sum of the proper divisors of a given number (i.e., excluding itself)</a>. 
</div>

```scala
def sumDivisors(n: Int): Int = 
    if n<=1 then 0
    else 
        var sum=0
        for i <- 1 to n/2 do
            if n%i==0 then sum+=i
        sum
```

```scala
assert(sumDivisors(0) == 0)
assert(sumDivisors(1) == 0)
assert(sumDivisors(10) == 8)
assert(sumDivisors(26) == 16)
assert(sumDivisors(100) == 117)
```

Use the `sumDivisors` function to test if two numbers are <a href="https://en.wikipedia.org/wiki/Amicable_numbers">amicable</a>, and to check wether a given number is <a href="https://en.wikipedia.org/wiki/Perfect_number">perfect</a>. 
</div>

```scala
//un numero amigable es aquel cuya suma de sus divisores es igual al otro numero y al reves
def isAmicable(n: Int, m: Int): Boolean = 
    if sumDivisors(n)==m then true
    else if sumDivisors(m)==n then true
    else 
        false
```

```scala
//otra forma de hacer el codigo mas reducido
def isAmicable2(n: Int, m: Int): Boolean =
  sumDivisors(n) == m && sumDivisors(m) == n
```

```scala
assert(isAmicable(284, 220))
assert(!isAmicable(16, 16))
assert(isAmicable(1210,1184))
assert(!isAmicable(1222, 333))
assert(isAmicable(2924,2620))
```

```scala
isAmicable(284, 220)
!isAmicable(16, 16)
isAmicable(1210,1184)
!isAmicable(1222, 333)
isAmicable(2924,2620)
```

```scala
//un numero es perfecto si la suma de sus divisores es igual a el mismo
def isPerfect(n: Int): Boolean = 
    if (sumDivisors(n)==n && n>0) true //se añade condicion de distinto de cero porque la funcion de divisors devuelve el cero tb
    else
        false
```

```scala
assert(!isPerfect(0))
assert(isPerfect(6))
assert(isPerfect(28))
assert(!isPerfect(27))
assert(isPerfect(496))
assert(isPerfect(8128))
```

```scala
!isPerfect(0)
isPerfect(6)
isPerfect(28)
!isPerfect(27)
isPerfect(496)
isPerfect(8128)
```
